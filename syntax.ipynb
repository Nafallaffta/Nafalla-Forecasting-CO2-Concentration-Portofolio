```python
import pandas as pd
import numpy as np
from statsmodels.tsa.stattools import acf

# --- Fungsi cek seasonality berdasarkan ACF ---
def detect_seasonality(series, max_lag=365, threshold=0.3):
    """Return lag dengan korelasi tertinggi > threshold, selain lag=0"""
    series = series.dropna()
    if len(series) < max_lag:
        return None, None

    acf_values = acf(series, nlags=max_lag, fft=True)
    # Abaikan lag 0
    lags = np.arange(1, len(acf_values))
    acf_values = acf_values[1:]

    # Cari spike yang melewati threshold
    max_corr = np.max(acf_values)
    lag = lags[np.argmax(acf_values)]

    if max_corr > threshold:
        return lag, max_corr
    else:
        return None, max_corr

# --- Loop semua kolom numerik ---
summary = []
num_cols = train.select_dtypes(include=["float64", "int64"]).columns.tolist()

for col in num_cols:
    lag, corr = detect_seasonality(train[col], max_lag=365, threshold=0.3)
    if lag is not None:
        summary.append({
            "variable": col,
            "seasonal": True,
            "lag_detected": lag,
            "corr": round(corr, 3)
        })
    else:
        summary.append({
            "variable": col,
            "seasonal": False,
            "lag_detected": None,
            "corr": round(corr, 3) if corr is not None else None
        })

# --- Jadiin DataFrame ---
seasonality_report = pd.DataFrame(summary)
print(seasonality_report)
```

```python
#Imputasi Seasons
def map_season(month):
    if month in [1, 2, 11, 12]:
        return "cool"
    elif month in [3, 4, 5]:
        return "hot"
    elif month in [6, 7, 8, 9, 10]:
        return "rainy"
    else:
        return None
```

```python
train['seasons'] = train['seasons'].fillna(train['month'].apply(map_season))
test['seasons'] = test['seasons'].fillna(test['month'].apply(map_season))
```

```python
#Imputasi Rain
def map_rain(precipprob):
    if precipprob == 100:
        return True
    else:
        return None  # biarkan MV tetap
```

```python
train['rain'] = train['rain'].fillna(train['precipprob'].apply(map_rain))
test['rain'] = test['rain'].fillna(test['precipprob'].apply(map_rain))
```

```python
def map_rain_conditions(cond):
    cond = str(cond).lower()  # pastikan huruf kecil
    if cond in ['rain, partially cloudy']:
        return True
    elif cond == 'clear':
        return False
    else:
        return None  # biarkan MV tetap jika kondisi lain
```

```python
train['rain'] = train['rain'].fillna(train['conditions'].apply(map_rain_conditions))
test['rain'] = test['rain'].fillna(test['conditions'].apply(map_rain_conditions))
```

```python
#Imputasi Conditions
def impute_conditions(row):
    if pd.isna(row['conditions']):
        if row['precipprob'] == 100:
            return 'rain, partially cloudy'
        elif row['precipprob'] == 0 and row['rain'] == True:
            return 'partially cloudy'
        else:
            return None  # biarkan MV jika kondisi lain
    else:
        return row['conditions']  # kalau sudah ada, biarkan
```

```python
train['conditions'] = train.apply(impute_conditions, axis=1)
test['conditions'] = test.apply(impute_conditions, axis=1)
```

### Outlier

```python
import pandas as pd
import numpy as np
from scipy.stats import zscore

# 1️⃣ Pilih kolom numerik (tanpa target)
numerik_cols = [col for col in train.select_dtypes(include=['number']).columns if col != 'daily_ktCO2']

# 2️⃣ Hitung Z-score untuk TRAIN
z_scores = train[numerik_cols].apply(zscore)
outliers = (np.abs(z_scores) > 3)
print("Jumlah outlier per kolom (train):")
print(outliers.sum())

# 3️⃣ Tampilkan beberapa baris yang ada outlier
outlier_rows = train[outliers.any(axis=1)]
print("Baris dengan outlier (train):")
print(outlier_rows.head())

# 4️⃣ Winsorization (1% dan 99%)
lower = train[numerik_cols].quantile(0.01)
upper = train[numerik_cols].quantile(0.99)

train_capped = train.copy()
for col in numerik_cols:
    train_capped[col] = np.clip(train[col], lower[col], upper[col])

# 5️⃣ Terapkan ke TEST pakai batas dari TRAIN
test_capped = test.copy()
for col in numerik_cols:
    test_capped[col] = np.clip(test[col], lower[col], upper[col])

# ✅ Optional: cek hasil
print("\nSebelum winsorization:")
print(train[numerik_cols].describe())
print("\nSesudah winsorization:")
print(train_capped[numerik_cols].describe())
```

```python
from scipy.stats import zscore
import numpy as np

# Hitung ulang Z-score setelah winsorization
z_scores_capped = train_capped[numerik_cols].apply(zscore)

# Tandai outlier (pakai threshold 3)
outliers_capped = (np.abs(z_scores_capped) > 3)

# Hitung jumlah outlier per kolom
outliers_count_capped = outliers_capped.sum()

print("Jumlah outlier setelah winsorization:")
print(outliers_count_capped)

# Total outlier keseluruhan
total_outlier_after = outliers_capped.any(axis=1).sum()
print(f"\nTotal baris yang masih mengandung outlier: {total_outlier_after}")
```

```python
# --- Definisikan variabel hasil feature engineering
after_groups = {
    "Temperature": ["temp_pca"] if "temp_pca" in X_train.columns else [],
    "Feelslike": ["feelslike_pca"] if "feelslike_pca" in X_train.columns else [],
    "Solar Energy": ["solarenergy"],  # setelah solarradiation dibuang
    "Combined TempFeels": ["tempfeels"] if "tempfeels" in X_train.columns else []
}

# --- Hitung korelasi tiap fitur hasil FE ke target
after_corr = []

for group_name, cols in after_groups.items():
    for col in cols:
        if col in X_train.columns:
            corr = np.corrcoef(X_train[col], Y_train)[0, 1]
            after_corr.append({"Kelompok": group_name, "Fitur": col, "Korelasi ke CO2": corr})

# --- Tampilkan tabel
after_corr_df = pd.DataFrame(after_corr)
after_corr_df = after_corr_df.sort_values(by="Korelasi ke CO2", ascending=False).reset_index(drop=True)

print("=== Korelasi Sesudah Feature Engineering ===")
display(after_corr_df)
```

### One Hot Encoding + Scaling

```python
import pandas as pd
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import MinMaxScaler, OneHotEncoder

# ---- Fungsi otomatis deteksi kolom ----
def get_col_types(df):
    num_cols = df.select_dtypes(include=["int64", "float64"]).columns.tolist()
    cat_cols = df.select_dtypes(include=["object", "category"]).columns.tolist()
    return num_cols, cat_cols

# ---- Pisahkan kolom ----
num_cols, cat_cols = get_col_types(X_train)

print("Kolom Numerik:", num_cols)
print("Kolom Kategorik:", cat_cols)
# Pastikan semua kolom kategorikal jadi string
for col in cat_cols:
    X_train[col] = X_train[col].astype(str)
for col in cat_cols:
    X_test[col] = X_test[col].astype(str)

# ---- Pipeline ----
num_pipeline = Pipeline(steps=[
    ('scaler', MinMaxScaler(feature_range=(0, 1)))  # default (0,1), bisa juga (-1,1)
])

cat_pipeline = Pipeline(steps=[
    ('ohe', OneHotEncoder(handle_unknown='ignore', drop='first', sparse_output=False))
])

# ---- Column Transformer ----
preprocessor = ColumnTransformer(
    transformers=[
        ('num', num_pipeline, num_cols),
        ('cat', cat_pipeline, cat_cols)
    ],
    remainder='drop'  # kolom selain num & cat dibuang
)

```

```python
X_train_transformed = preprocessor.fit_transform(X_train)
X_test_transformed = preprocessor.transform(X_test)

print("Shape X_train_transformed:", X_train_transformed.shape)
print("Shape X_test_transformed:", X_test_transformed.shape)
```

```python
#jadikan dataframe
# --- ambil nama kolom numerik (tetap sama) ---
num_features = num_cols

# --- ambil nama kolom kategorik hasil OHE ---
cat_features = preprocessor.named_transformers_['cat']\
    .named_steps['ohe']\
    .get_feature_names_out(cat_cols)

# --- gabung semua nama kolom ---
all_features = list(num_features) + list(cat_features)

# --- jadi DataFrame biar enak lihatnya ---
X_train_transformed = pd.DataFrame(X_train_transformed, columns=all_features)

print(X_train_transformed.head())
print("Total kolom:", len(all_features))
```
